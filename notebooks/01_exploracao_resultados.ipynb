```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import polars as pl
import seaborn as sns

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
plt.rcParams["figure.dpi"] = 120
plt.style.use("seaborn-v0_8-whitegrid")
```

## Base municipal e tipologias

```python
clusters = pl.read_parquet(ROOT / "data" / "gold" / "municipios_clusters.parquet")
clusters.select("cluster").to_series().value_counts().sort("cluster")
```

## IDEB por tipologia

Distribuição do IDEB (anos iniciais) em cada cluster: o IDEB **não** entrou na clusterização.

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
sns.boxplot(
    data=clusters.drop_nulls(["cluster", "ideb_early_years"]).to_pandas(),
    x="cluster",
    y="ideb_early_years",
    ax=ax,
)
ax.set_title("IDEB (anos iniciais, rede pública) por tipologia")
ax.set_xlabel("Cluster")
ax.set_ylabel("IDEB (nota de 0 a 10)")
plt.tight_layout()
```
